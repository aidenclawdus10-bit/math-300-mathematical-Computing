### Problem 2
We want to compute $\int_0^1 x^2dx$. To do so, generate random points between 0 and 1. Accept them if they are below and reject them if they are above. Compute the average of the accepted values and plot the accepted vs rejected points.

```python
X = np.linspace(0,1,1000)
plt.plot(X,X**2)
plt.title("x^2")
plt.xlim([0,1])
plt.ylim([0,1])
def p(n):
    inside_count=0
    x=np.random.uniform(0,1,n)
    y = np.random.uniform(0,1,n)
    for i,j in zip(x,y):
            if i**2>=j:
                inside_count += 1
    print(inside_count)
    return(inside_count/n)
print(p(100))
```

### Problem 4

Estimate $\pi$ using one of the functions from lecture with:

- 1,000 samples
- 10,000 samples
- 100,000 samples

Compare the results with the true value of $\pi$.

```python
l=5
def p(n):
    inside_count=0
    x=np.random.uniform(-l,l,n)
    y=np.random.uniform(0,1,n)
    for i,j in zip(x,y):
            if np.exp(-i**2)>=j:
                inside_count += 1
    print(inside_count)
    return((2*l*inside_count/n)**2)
print(p(1000))
print(p(10000))
print(p(100000))
```
